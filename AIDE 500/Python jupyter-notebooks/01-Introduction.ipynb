## Objects and Types
In a Python program, elements are 'objects'. Each object has a specific data-type.

Python guesses the data-type from the object initialization. Let's create a variable `x` that stores the integer `5` and experiment with it.

```python
x = 5
type(x)
```

```python
x = 3.14
type(x)
```

```python
x = True
type(x)
```

```python
x = "CSC243"
type(x)
```

There are two main object types:

1. Scalar, cannot be decomposed: `int`, `float`, `bool`, and `NoneType` (seen later)

2. Non-Scalar, have internal structure: `string`, `list`, `tuple`, `dict` (last 3 seen later)

### The int type
The `int` type is used to represent integers and has unlimited precision (bounded by memory restrictions of course). Example integers include:
- negative numbers such as: -5, -71, etc.
- non-negative numbers such as: 0, 12, 93801, etc.

```python
n = 313131000849184902084082114801
m = -n*n
print(m)
```

### The float type
The `float` is used to represent real numbers, up to a certain precision (`float` stands for floating point number).

Note here that scientific notation can be used to express floats: 2870 = 2.87 * 1000 = 2.87E3

```python
first_number = 2.87E3
print(first_number)
```

```python
second_number = 5.0
print(second_number)
```

```python
their_sum = first_number + second_number
print(their_sum)
```

```python
print(their_sum*2)
```

The binary representation of floats uses 64 bits, hence 8 bytes of memory: 1 bit is for the sign, 52 bits are the significant part (a.k.a the mantissa), 11 remaining bits to represent the exponent.

### The bool type
Used to represent Boolean/logical values: `True` and `False`. In Python, `True` and `False` are keywords (i.e., cannot be used as variable names).

Boolean values originate from logical tests, such as the below code:

```python
a = (8>3)
print(a)
```

```python
b = (2>3)
print(b)
```

```python
c = not b
print(c)
```

## Arithmetic Operators
The arithmetic operators for the `int` and `float` types include the following: Addition (`+`), Subtraction (`-`), Multiplication (`*`), Division(`/`), Power(`**`).

```python
x = 12.1
y = 4
```

```python
z = x + y
print(z)
```

```python
z = x - y
print(z)
```

```python
z = x*y
print(z)
```

```python
z = x/y
print(z)
```

```python
z = x**y
print(z)
```

Except for division `/`, if both `x` and `y` are of type `int`, the result is an `int`.  In all other cases, the result is a `float`.

All above are **binary** arithmetic operators: they take in **two** operands. There exists the **unary** operator minus `-`: `-x` is the negative of `x`.

There are also operators specific to the `int` data-type: 

- integer division `//`: keeps the integer part of the division of `a//b`

- modulo `%`: returns the remainder for the division of `a/b`. Note that it returns an integer which is at least `0` and at most `b-1`. The modulo operator `%` does NOT return the decimal part of said division.

Let's try them out.

```python
x = 12 #x is set to an int
```

```python
z = x//y
print(z)
```

```python
y = 5
z = x % y # x&y is the remainder of x/y
print(z)
```

```python
y = 3 
z = x % y 
print(z)
```

## Relational Operators
Relational operators are binary operators that evaluate the relationship between two inputs, and output a truth value (i.e., `bool`). For example, the `<` operator evalues if a number is less than another. The `==` operator checks if two operands are the same. The `!=` operator checks if two operands are different.

Below are the possibilites.

```python
print(z == y)
```

```python
print(z < y)
```

```python
print(z <= y)
```

```python
print(z > y)
```

```python
print(z != y) #not equal check
```

## Logical (`bool`) Operators
Logical operators accept booleans only as their inputs and -- like relational operators -- return a boolean. The general format is: `x` `operator` `y` --> `z`; where `x`, `y`, and `z` are all boolean.

```python
x = True
y = False
```

```python
z = x and y
print(z)
```

```python
z = x or y
print(z)
```

```python
z = not y
print(z)
```

### Order of precedance
In Python, some operators take precedence over others (i.e., they get executed before others). 

In general, a good way to remember the precedence of operators is to memorize the PEMDAS acronym. PEMDAS stands for Parenthesis, Exponentiation, Multiplication, Division, Addition, Subtraction. Accordingly, precedence is in this order. 

In Python, PEMDAS is followed by 'Relational operators', `not`, `and`, and finally the `or` operator.

Some examples below:

The following expression:

```python
-2*3/5+4*5
```

is interpreted as:

```python
(((-2)*3)/5) +(4*5)
```

The following expression:

```python
x>=1 and x<10 or x>=20 and x<20
```

is interpreted as:

```python
((x>=1) and ( x<10)) or ((x>=20) and(x<20))
```

## Output to the console, using `print()`
As we have already seen, to output to the console the contents of variables we make use of the built-in Python function `print()`. For example, we have in the below a variable called `y` with contents equal to the float 5.0.

```python
y = 5.0
print(y)
```

The `print()` function takes as input a variable name as in the above, and outputs the value as a string. Alterntively, `print()` can take in as input a string, and output a string. For example

```python
print("This string will be printed")
```

Note how the quotations in the string were not displayed; just the actual sequence of characters. If we want our string to contain quotations, we need to make use of the special backslash character `\` and follow it with single or double quotations. Here, we have:

```python
string_with_quotations = "\"This string contains double quotations\"" 
print(string_with_quotations)
```

```python
string_with_quotations = "\'This string contains single quotations\'" 
print(string_with_quotations)
```

There are other special characters, called escape sequences, that are especially useful for formatting a string to our taste. For example, if we want to go to a newline, we use the `\n` escape sequence. To insert a tab, we use the `\t` escape sequence.

Some examples:

```python
my_string = "This is line one.\nThis is line two." #notice the effect of the space between \n and This
print(my_string)
```

```python
my_string = "This\tsentence\tis\ton\ta\tsingle\tline."
print(my_string)
```

The built-in `print()` function can also take multiples inputs. The result will be printed using the default seperator (a white-space ` `) and default line-ending character (a newline `\n`). For example, if we have 3 variables with different contents:

```python
var1 = "The speed is equal to"
var2 = 56.1
var3 = "meters per second."
print(var1, var2, var3)
```

```python
x = 1.0
y = 2.3
print("The variable x =", x, "and y =", 2.3)
print("The variable x = " + str(x) + " and y = " + str( 2.3))
```

Notice how `var2`, `x`, and `y` -- whose contents are `float` -- were converted to `str` and inserted in their proper order. 

Also, note how a space ` ` is separating the contents of `var1`, `var2`, and `var3`. We can change the default separator ` ` to **any other string** by specifying the separator parameter as in the below:

```python
var1 = "Lebanese"
var2 = 'American'
var3 = "University"
print(var1, var2, var3)
```

```python
print(var1, var2, var3, sep = "\t")
```

```python
print(var1, var2, var3, sep = "*****")
```

```python
print(var1, var2, var3, sep = "-.-.-.-")
```

```python
print(var1, var2, var3, sep = " | ")
```

Finally, we can specify the `end` parameter for `print` as well. By default, it is set to `\n` and consecutive prints will be each displayed on a new line.

```python
print("first")
print("second")
print("third")
print("fourth")
```

```python
print("first", end="\t")
print("second", end="\n")
print("third", end="\t")
print("fourth")
print("fifth", end="")
print("\tsixth")
```

```python
height = input("Please enter your height (in cms): ")
```

```python
print(height)
```

```python
type(height)
```

```python
type(height)
```

```python
type(height)
```

The conversion to integer is necessary here since we might need to perform calculations using the input value later.

## Our first program: Swapping two values
This first program will illustrate input from the keyboard, variables, escape characters, casting, and output.

```python
a = input("Enter the first integer: ")
a = int(a)
b = input("Enter the second integer: ")
b = int(b)
## perform swapping, need a temporary variable
temp = a
a = b
b = temp
print("\n")
print("The first integer is now:", a)
print("The second integer is now:", b)
print("\nThis concludes our first code.")
```

## Importing modules
Often times, there are functions that we might need to import from modules. Such examples include mathematical functions like `cos`, `sin`, `exp`, etc. Other examples include modules that allow for plotting.

In the below code, we will examine the `math` module. First, we import it:

```python
import math
```

Next, say we need to use square-root function `sqrt`. We must specify to Python that it is being retrieved from the `math` module by using the member access `.` operators; as in the below: 

```python
x = float(input("Enter an integer: "))
y = math.sqrt(x)
print("The square-root of", x, "is", y)
```

Alternatively, we can import **just** the `sqrt` method by typing in:

```python
from math import sqrt
```

Now, we don't need to `math.` anymore to use it:

```python
x = float(input("Enter an integer: "))
y = sqrt(x)
print("The square-root of", x,"is", y)
```
