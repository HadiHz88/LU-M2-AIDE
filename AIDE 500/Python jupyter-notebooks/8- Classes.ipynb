```python
class Point(object):
    def __init__(self, x_value=0, y_value=0):
        self.x = x_value
        self.y = y_value
        
    def __str__(self):
        return "(" + str(self.x) + "," + str(self.y) + ")"
    
a = Point()
print(a)
b = Point(3, 5)
print(b)
```

```python
from math import pi

class Circle(object):    
    def __init__(self, c=Point(0,0), r=1):
        self.center = c
        self.radius = r
    
    def diameter(self):
        return 2*self.radius
    
    def perimeter(self):
        return 2*pi*self.radius
    
    def area(self):
        return pi*self.radius**2
    
```

```python
C = Circle()
print(C.diameter())
print(C.perimeter())
print(C.area())

```
