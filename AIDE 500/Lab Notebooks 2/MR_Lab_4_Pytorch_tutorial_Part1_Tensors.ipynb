# 1. Tensors
--------------------------------------------

Tensors are a specialized data structure that are very similar to arrays
and matrices. In PyTorch, we use tensors to encode the inputs and
outputs of a model, as well as the model’s parameters.

Tensors are similar to NumPy’s ndarrays, except that tensors can run on
GPUs or other specialized hardware to accelerate computing. If you’re familiar with ndarrays, you’ll
be right at home with the Tensor API. If not, follow along in this quick
API walkthrough.


```python
import torch
import numpy as np
```

```python
print(torch.__version__)
```

### Tensor Initialization


Tensors can be initialized in various ways. Take a look at the following examples:

**Directly from data**

Tensors can be created directly from data. The data type is automatically inferred.



```python
data = [[1, 2],[3, 4]]  # data is a 2D list
x_data = torch.tensor(data)
print (x_data)
```

**From a NumPy array**

Tensors can be created from NumPy arrays (and vice versa - see *bridge-to-np-label*).



```python
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
print("data:" , data)
print("np_array:" , np_array)
print ("x_np:" , x_np)
```

**From another tensor:**

The new tensor retains the properties (shape, datatype) of the argument tensor, unless explicitly overridden.



```python
x_ones = torch.ones_like(x_data) # retains the properties of x_data (shape and datatype)
print(f"Ones Tensor: \n {x_ones} \n")

x_rand = torch.rand_like(x_data, dtype=torch.float) # overrides the datatype of x_data and keeps the shape
print(f"Random Tensor: \n {x_rand} \n")
```

**With random or constant values:**

*shape* is a tuple of tensor dimensions. In the functions below, it determines the dimensionality of the output tensor.



```python
shape = (2,3)  # shape is tuple
rand_tensor = torch.rand(shape)
rand_tensor2 = torch.rand((2,3))
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Random Tensor2: \n {rand_tensor2} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

### Tensor Attributes


Tensor attributes describe their shape, datatype, and the device on which they are stored.



```python
tensor = torch.rand(3,4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

### Tensor Operations


Over 100 tensor operations, including transposing, indexing, slicing,
mathematical operations, linear algebra, random sampling, and more are
comprehensively described [here](https://pytorch.org/docs/stable/torch.html).

Try out some of the operations from the list.
If you're familiar with the NumPy API, you'll find the Tensor API a breeze to use.



**Standard numpy-like indexing and slicing:**



```python
tensorA = torch.ones(4, 4)
print(tensorA, "\n")

tensorA[1,1] = 6
print(tensorA, "\n")

tensorA[1,3] = 4
print(tensorA, "\n")

tensorA[:,2] = 3
tensorA[:,3] = 5
print(tensorA, "\n")

tensorA[:,:] = 2
print(tensorA, "\n")

tensorA[3,:] = 0
print(tensorA)
```

**Joining tensors** You can use *torch.cat* to concatenate a sequence of tensors along a given dimension.

tensors: A sequence (such as a list or tuple) of PyTorch tensors.

dim: The dimension along which to concatenate them (defaults to 0).

out: An optional output tensor to store the result. 

(https://www.codecademy.com/resources/docs/pytorch/tensors/cat), (https://docs.pytorch.org/docs/stable/generated/torch.stack.html)

See also [torch.stack] (https://pytorch.org/docs/stable/generated/torch.stack.html),
another tensor joining operation that is subtly different from *torch.cat*.



```python
t1 = torch.cat([tensorA, tensorA, tensorA], dim=0)
print(t1)
print(t1.shape)
```

```python
t1 = torch.cat([tensorA, tensorA, tensorA], dim=1)
print(t1)
print(t1.shape)
```

```python
# Shape (2, 3)
x = torch.tensor([[1, 2, 3], 
                  [4, 5, 6]])

# Concatenate along rows (dim=0) -> Result shape becomes (4, 3)
cat_dim0 = torch.cat([x, x], dim=0)
print("Concatenated along dim 0:\n", cat_dim0)

# Concatenate along columns (dim=1) -> Result shape becomes (2, 6)
cat_dim1 = torch.cat([x, x], dim=1)
print("\nConcatenated along dim 1:\n", cat_dim1)
```

**Multiplying tensors**



```python
print(tensorA)
# This computes the element-wise product
print(f"tensorA.mul(tensorA) \n {tensorA.mul(tensorA)} \n")
# Alternative syntax:
print(f"tensorA * tensorA \n {tensorA * tensorA}")
```

This computes the matrix multiplication between two tensors



```python
print(f"tensorA.matmul(tensorA.T) \n {tensorA.matmul(tensorA.T)} \n")
# Alternative syntax:
print(f"tensorA @ tensorA.T \n {tensorA @ tensorA.T}")
```

**In-place operations**
Operations that have a `_` suffix are in-place. For example: `x.copy_(y)`, ``x.t_()``, will change ``x``.



```python
print(tensorA, "\n")
tensorA.add_(5)
print(tensorA)
```

```python
print(tensorA, "\n")
tensorA.sub_(3)
print(tensorA)
```

```python
print(tensorA, "\n")
tensorA.t_()
print(tensorA)
```

```python
print(tensorA, "\n")
tensorA.copy_(5)
print(tensorA)
```

**Bridge to NumPy**

Tensor to NumPy array




```python
t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```

A change in the tensor reflects in the NumPy array.



```python
t.add_(2)
print(f"t: {t}")
print(f"n: {n}")
```

NumPy array to Tensor



```python
n = np.ones(5)
t = torch.from_numpy(n)
print(t)
```

Changes in the NumPy array reflects in the tensor.



```python
np.add(n, 1, out=n)
print(f"t: {t}")
print(f"n: {n}")
```
