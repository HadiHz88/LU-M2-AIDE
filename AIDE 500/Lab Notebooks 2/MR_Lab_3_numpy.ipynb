# **The `numpy` module**
## Introduction
[Numpy](https://numpy.org) is a Python package, it stands for numerical python. Using `numpy` we can perform operations such as:
- Mathematical and logical operations on arrays
- Operations related to linear algebra
- Built-in functions for random number generation

```python
import numpy as np
```

## The `ndarray` object
The most important object defined in NumPy is an N-dimensional array type called `ndarray`:
- It describes the collection of items of the same type
- Items in the collection can be accessed using a zero-based index

```python
L = [1, 2, 3]
print(L)
a = np.array(L)
print(a)
```

```python
L = [[1, 2, 3], [4, 5, 6]]
a = np.array(L)
print(a)
```

Computing the length of the array in `numpy` is done via the `shape` attribute:

```python
print(a.shape)
```

Unlike Python lists, you can specify the data-type of the items in a `numpy` array. Python automatically casts all the items to the specified `dtype`. Once you specify the `dtype`, mixed types will not be allowed.

```python
a_mixed = np.array([1, 'CSC243'])
print(a_mixed)
a_float = np.array([1, 2, 3 ], dtype=float)
print(a_float)
a_float = np.array([1.5, 2.7, 3.1 ], dtype=int)
print(a_float)
```

### Creating `ndarray` objects
Numpy `ndarrays` are created:
- As empty : empty function allocates memory without clearing it. The values inside will be arbitrary "**garbage data**" left over in your computer's RAM. You must overwrite them before reading from the array.
- As zeros
- As ones
- From existing data

```python
a = np.empty((3,4), dtype=float)
print(a)
a = np.zeros((2,4))
print(a)
a = np.ones((3,2))
print(a)
a = np.ones((3,), dtype=int)
print(a)
```

### Ranges
You can create ranges of data through: 
- `np.arange(start, stop, step, dtype)`: returns an `ndarray` object containing evenly spaced values within a given range.
- `np.arange(stop, dtype)`: returns an `ndarray` object containing evenly spaced values from 0 till stop-1.
- `np.linspace(start, stop, num, dtype)`: instead of step size, the number **num** of evenly spaced values in the interval is specified.

```python
r = np.arange(1.5, 2, 0.1)
print(r)
r = np.arange(5)
print(r)
r = np.linspace(10, 50, 5)
print(r)
```

One advantage of using `numpy.arange` instead of range is that the step-size can
be of type `float` in this case.

### Math functions using `numpy`
Mathematical functions can also be accessed from the `numpy` module (sin,
cos, exp, ceil, floor, etc.). The advantage of using `numpy` functions is that we can pass a full array as an input (in contrast to the functions from the `math` module, which take a single value as input).

Example: plotting exp(-x)*cos(2πx), for x = [0, 0.1, 0.2, …, 9.9]

```python
from math import pi
import matplotlib.pyplot as plt

X = np.arange(0, 10, 0.1)
Y = np.exp(-X)*np.cos(2*pi*X)
print(Y)
plt.plot(X, Y)
```

### Statistical Operations
We can compute the minimum and maximum through `np.amin(a, axis=)` and `np.amax(a, axis=)`. The `axis` keyword specifies the dimension over which min/max is computed. If not specified, the min/max will be computed over the whole array.

```python
a = np.array([[7,2], [3,4], [5,6]])
print("a =\n" + str(a))
print("min:", np.amin(a))
print("minima (axis=0):", np.amin(a, axis=0))
print("minima (axis=1):", np.amin(a, axis=1))
```

We can also compute the average (mean), median, standard deviation and variance. We use in these cases `np.mean()`, `np.median()`, `np.std()`, and `np.var()`, respectively. All these functions also admit an `axis` keyword parameter like `np.amin`.

### Generation Random Numbers
We can generate random numbers using `numpy.random`. Here are a few examples:

```python
import numpy.random as rand

x = rand.uniform(1, 10)
print(x)
```

The `random.uniform(x, y)` function generates a uniformly-distributed random `float` in the real interval [x, y]. The default values are `x=0, y=1`.

The `random.randint` generates a random integer between `a` and `b` inclusive.
Multiple runs of the same code give different random numbers (that is whole point).

```python
a = np.random.randint(3,500)
print("a=",a)
```

```python
a = 1
b = 5
y = rand.randint(a, b+1)
print(y)
```

```python

arrayD = np.random.randint(3,500,10)
print(arrayD.shape)
print ("arrayD=" , arrayD)
```

## Pseudo-random values
Numpy can be used to generate pseudo-random values from various distributions. 

In particular, a very useful distribution is the standard normal (zero mean and unit variance).

Let's generate two vectors sampled from the normal distribution, using a length parameter that we'll be able to change if needed.

The **numpy.random.randn()** function in Python's NumPy library is used to generate random numbers sampled from the standard normal distribution.

```python
length = 50

vecA = np.random.randn(length)
vecB = np.random.randn(2*length)

print("vecA:" , vecA)
print("vecB:" , vecB)
```

vecA and vecB are numpy *arrays*. One of their attributes can be fetched to check their *shape*

```python
print(vecA.shape)
print(vecB.shape)
```

Numpy arrays can be vectors as well as matrices, or any tensor. For example the following code will create tensors with 3 dimensions using the standard normal

```python
arrayC = np.random.randn(3,10,4)
print(arrayC.shape)
print(arrayC)
```

A very important features of arrays is the fact they can be used as *iterables*. For example, you can iterate over the dimensions of an array by simply "looping" over it using a *for* loop

```python
for dim in arrayC:
    print(dim.shape)
```

Also possible to enumerate along the dimension in order to get the index of the current "smaller" array

```python
print('Initial shape is %d %d %d' % (arrayC.shape[0],arrayC.shape[1],arrayC.shape[2]))
print('Iterating over the first dimension using an index k')
for k,curdim in enumerate(arrayC):
    print('k = %d, shape is %d %d' % (k,curdim.shape[0],curdim.shape[1]))
```

Another important operation when working with Numpy Arrays is *reshaping*. Essentially, *reshaping* consists in changing the organisation of the array (in terms of dimension), while keeping the same number of elements. For example, a 20x10 2D array can be converted into a 4x5x10 array

```python
A = np.random.randint(1,50,(10,20))
print('Initial shape of A is %d x %d' % (A.shape[0],A.shape[1]))
print(A)
B = A.reshape((4,5,10))
print('B is A reshaped to %d x %d x %d' % (B.shape[0],B.shape[1],B.shape[2]))
print(B)
```

One property of numpy that is really important is broadcasting. The goal of broadcasting is to simplify the vectorization of certain operations when the vectors do not have the same shape. For example you can easily perform element-wise multiplication.


```python
x = np.array([2,3])
y = np.array([[4,1],[9,10],[12,13]])
result = x*y
print("X: ",x)
print("Y: ")
print(y)
print("X shape is: ",x.shape)
print("Y shape is: ",y.shape)
print("Element-wise multiplication shape:", result.shape)
print("Element-wise multiplication result:")
print(result)
```

Another very powerful tool in numpy is indexing. You can use either an integer vector or a boolean vector to choose which indexes you want to extract from your numpy tensor.

Consider that we want to extract all elements from the first line of your vector y that have a higher value than 1, you would have to do:

```python
first_row = y[0]
print(first_row)
first_row_higher_than_one = first_row > 1
print("Result: ", first_row[first_row_higher_than_one])
```

You can also choose specific lines to query, for example if you want to query lines 0 and 2

```python
rows = [0,2]
rows_result = y[rows]
values_higher_than_one = rows_result > 1
print("Result: ", rows_result[values_higher_than_one])
```

You can also save and load your numpy tensors using np.savez and np.load. This will be really important in the next courses as this enable you to generate your data only once instead of having to do all the calculations every time you need your data.

```python
filename = "y.npz"
source_tensor = y
np.savez(filename,data=source_tensor)
```

```python
loaded_npz = np.load(filename)
loaded_tensor = loaded_npz["data"]
print("Your tensor was loaded and contains: ", loaded_tensor)
```
