## Exceptions
Python exceptions occur whenever the interpreter cannot execute a line of code. We have seen many kind of exceptions so far, such as `TypeError`, `NameError`, and `ValueError` exceptions. It is normal to encounter errors while programming, but a programmer should write code that handles exceptions well in case they occur. Handling exceptions makes sure that the program does not crash in unfamiliar cases.

## The `try-except` structure
The `try-except` structure allows handling specific or general exceptions when they occur. Let's examine an example that might result in an exception:

```python
x = int(input("Enter an integer: "))
y = int(input("Enter an integer: "))

z = x/y
print(z)
```

Here, we might get either:
- `ZeroDivisionError` : if y == 0.
- `ValueError`: if user enters non-integer values.

The `try-except` structure allows to handle previous errors by:
- Placing our code inside a `try` block
- Writing `except` blocks for each expected error

```python
try:
    x = int(input("Enter an integer: "))
    y = int(input("Enter an integer: "))

    z = x/y
    print(z)

except ZeroDivisionError:
    print("Cannot divide by 0!")
except ValueError:
    print("Bad input! Please enter integers.")
```

In addition, we can avoid writing `except` blocks for each error if we want to handle all errors at once. We do so by writing:

```python
try:
    x = int(input("Enter an integer: "))
    y = int(input("Enter an integer: "))

    z = x/y
    print(z)

except:
    print("Bad Input!")
```

Here, the `except` block handles all types of exceptions but prints the same message. Hence there is a tradeoff between lines of code written and detail of the display message.

### Optional clauses in `try-except`
We can end a `try-except` block with an optional `else` clause, the statemens under which get executed **only if no exceptions occurs**. This optional `else` clause is placed after all `except` clauses. Example:

```python
try:
    x = int(input("Enter an integer: "))
    y = int(input("Enter an integer: "))
    z = x/y
except:
    print("Bad Input!")
else:
    print(z)
```

We can end a `try-except` block with an optional `finally` clause, the statemens under which **always get executed**, whether an exception is raised or not. This optional `finally` clause is placed after all `except` clauses. Example:

```python
try:
    a = input("Enter an integer: ")
    x = int(a)
    b = input("Enter an integer: ")
    y = int(b)
    
    z = x/y
except:
    print("Bad Input!")
else:
    print(z)
finally:
    print("Original inputs:", a, b)
```
